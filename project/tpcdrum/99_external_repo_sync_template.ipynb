# 외부 AT_TPC-with-lilak 동기화 템플릿

현재 환경에서 GitHub 직접 clone이 막힐 수 있어서, 로컬에 받은 소스를 기준으로 복사하는 템플릿입니다.

## 사용 방법
1. `AT_TPC-with-lilak` 저장소를 로컬에 내려받아 이 환경에 업로드
2. 아래 경로를 실제 위치로 수정
3. `tpcdrum/`, `source/` 중 필요한 것만 복사

```python
from pathlib import Path
import shutil

# TODO: 실제 업로드된 경로로 수정
UPSTREAM = Path('/workspace/AT_TPC-with-lilak')
DST = Path('/workspace/test-repo/project/tpcdrum/upstream_snapshot')

DST.mkdir(parents=True, exist_ok=True)
```

```python
# tpcdrum 폴더 복사
src_tpcdrum = UPSTREAM / 'tpcdrum'
dst_tpcdrum = DST / 'tpcdrum'

if src_tpcdrum.exists():
    if dst_tpcdrum.exists():
        shutil.rmtree(dst_tpcdrum)
    shutil.copytree(src_tpcdrum, dst_tpcdrum)
    print('copied:', src_tpcdrum, '->', dst_tpcdrum)
else:
    print('missing:', src_tpcdrum)
```

```python
# source 폴더 복사 (선택)
src_source = UPSTREAM / 'source'
dst_source = DST / 'source'

if src_source.exists():
    if dst_source.exists():
        shutil.rmtree(dst_source)
    shutil.copytree(src_source, dst_source)
    print('copied:', src_source, '->', dst_source)
else:
    print('missing:', src_source)
```
